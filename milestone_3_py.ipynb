```python
# Define the word to be guessed
word_to_guess = "apple"

# Step 1: Define check_guess function
def check_guess(guess):
    # Step 2: Convert guess to lowercase
    guess = guess.lower()

    # Step 3: Check if the guess is in the word
    if guess in word_to_guess:
        print(f"Good guess! {guess} is in the word.")
    else:
        print(f"Sorry, {guess} is not in the word. Try again.")

# Step 1: Define ask_for_input function
def ask_for_input():
    while True:
        # Ask the user to guess a letter
        guess = input("Please guess a letter: ")

        # Step 2: Check that the guess is a single alphabetical character
        if len(guess) == 1 and guess.isalpha():
            # Step 3: Call check_guess to see if the guess is in the word
            check_guess(guess)
            break  # Exit the loop after a valid guess
        else:
            # If input is invalid, print error message
            print("Invalid letter. Please, enter a single alphabetical character.")

# Step 4: Call the ask_for_input function to test the code
ask_for_input()


```
