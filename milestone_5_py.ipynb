```python
# Import the required classes and functions from milestone_4
from milestone_4 import Hangman

def play_game(word_list):
    # Step 1: Initialize the number of lives and create an instance of the Hangman class
    num_lives = 5
    game = Hangman(word_list, num_lives)

    # Step 2: Start the game loop
    while True:
        # Check if the player has lost all their lives
        if game.num_lives == 0:
            print('You lost!')
            break

        # Check if there are still letters to guess
        if game.num_letters > 0:
            # Continue the game by asking for input
            game.ask_for_input()
        else:
            # If no letters remain, the player has won
            print('Congratulations. You won the game!')
            break
```
